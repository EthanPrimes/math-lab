# Prime Generator

Working to make a function that dynamically returns 1 prime at a time.

```python
from time import perf_counter as pf
```

```python
def prime_generator_fast():
    """Returns a generator which yields primes.

    Stores all previously computed primes and checks divisibility
    only by those to speed up the computation. Runs faster than
    prime_generator_efficient, but uses an increasing amount
    of space.

    Args:
        None

    Yields:
        int: the next prime

    Examples:
        >>> gen = prime_generator_faster()
        >>> next(gen)
        2
        >>> next(next(gen))
        5
    """
    yield 2
    primes = [2]
    n = 3
    while True:
        primeq = True
        for p in primes:
            if p * p > n:
                break
            if n % p == 0:
                primeq = False
                break
        if primeq:
            primes.append(n)
            yield n
        n += 2
```

```python
def prime_generator_efficient():
    """Returns a generator which yields primes.

    Uses trial division to determine primality. Uses minimal
    storage space but is slower than prime_generator_fast.

    Args:
        None

    Yields:
        int: the next prime

    Examples:
        >>> gen = prime_generator_efficient()
        >>> next(gen)
        2
        >>> next(next(gen))
        5
    """
    yield 2
    def trial_division_is_prime(n):
        if n <= 3:
            return n > 1
        for i in range(5, round(pow(n, 1/2)) + 1):
            if n % i == 0:
                return False
        return True
    n = 3
    while True:
        if trial_division_is_prime(n):
            yield n

        n += 2
```

```python
gen = prime_generator_fast()
print(next(gen))
print(next(gen))
```

```python
gen = prime_generator_fast()
for _ in range(25):
    print(next(gen))
```

```python
gen = prime_generator_efficient()
print(next(gen))
print(next(gen))
```

```python
gen = prime_generator_fast()
for _ in range(25):
    print(next(gen))
```

```python
# Test how long it takes to generate N primes.
# Determine the space taken up as well.
N = 10**6

# prime_generator_efficient
start = pf()
gen = prime_generator_efficient()
for _ in range(N):
    next(gen)
end = pf()
print("prime_generator_efficient")
print(f"time: {end - start}")

# prime_generator_fast
start = pf()
gen = prime_generator_fast()
for _ in range(N):
    next(gen)
end = pf()
print("prime_generator_fast")
print(f"time: {end - start}")
```
