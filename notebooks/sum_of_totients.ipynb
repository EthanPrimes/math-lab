Goal: Compute $\sum_{k=1}^n \phi(k)$ as quickly as possible

```python
from time import perf_counter as pf

from tqdm import tqdm

from core.primes import prime_factors, sieve_of_eratosthenes
```

```python
def euler_totient(n: int) -> int:
    if not isinstance(n, int):
        raise TypeError(f"Expected int, got {type(n).__name__}")
    if n < 1:
        raise ValueError(f"Expected positive integer, got {n}")

    primes = prime_factors(n).keys()
    res = n
    for p in primes:
        res = res // p * (p - 1)

    return res
```

```python
for n in tqdm(range(1, 10**4 + 1)):
    euler_totient(n)
```

```python
start = pf()
res = sieve_of_eratosthenes(10**6)
print(f"Time: {pf() - start}")
```
