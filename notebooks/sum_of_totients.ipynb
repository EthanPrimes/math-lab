```python
def fast_euler_totient(n):
    result = n
    p = 2
    while p * p <= n:
        # Check if p is a prime factor
        if n % p == 0:
            # If yes, update result using Euler's product formula
            while n % p == 0:
                n //= p
            result -= result // p
        p += 1

    # If n still has a prime factor greater than sqrt(n)
    if n > 1:
        result -= result // n
    return result


for i in tqdm(range(2, 10**5)):
    euler_totient(i)

for i in tqdm(range(2, 10**5)):
    fast_euler_totient(i)
```
