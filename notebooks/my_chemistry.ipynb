# My Chemistry

Every positive integer is an atom. Each atom can form a number of bonds up to its value. 1-1 and 1-2-1 are valid bond sets, but 1=1 is not.

```python
from collections import defaultdict
import random
```

```python
class Atom():
    def __init__(self, value=1):
        self.value = 1
        self.bonds = defaultdict(int)

    def update_bonds(self, atom_dict):
        updated_bonds = defaultdict(int)
        for atom, bonds in atom_dict.items():
            self.bonds[atom] += bonds

        print(self.bonds)
        if sum(self.bonds.values()) > self.value:
            raise ValueError("Made too many bonds.")
        elif any(val < 0 for val in self.bonds.values()):
            raise ValueError("Created negative bonds between atoms.")

        # Bonding the other atoms
        for atom, bonds in atom_dict.items():
            atom.bonds[self] += bonds

        self.bonds = updated_bonds
```

```python
atom1 = Atom()
atom2 = Atom()
atom1.update_bonds({atom2: 1})

print(atom1.bonds)
print(atom2.bonds)
```
