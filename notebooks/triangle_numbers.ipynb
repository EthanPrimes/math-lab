# Purpose
Examining triangle numbers, in particular trying to prove that every integer is a sum of three of them.
- We begin by examining writing triangle numbers as sums of triangle numbers.

```python
import math
```

```python
def tri(n):
    # Returns T_n
    return int(n*(n + 1)/2)

def tri_index(t):
    # Given T_n, returns n
    return math.isqrt(2 * t)

def is_tri(t):
    # Verifies that t is triangular
    index = tri_index(t)
    return tri(index) == t
```

```python
def ways_write_as_sum_triangles(n):
    # Determine all ways to write n as a sum of T_n
    triangles = [tri(i) for i in range(1, tri_index(n) + 1)]
    triangles = [t for t in triangles if t <= n]
    print(triangles)
```

```python
def count_sum_of_tri(n, tri_list):
    if n in (0, 1):
        return 1
    elif n < 0:
        return 0
    return sum(count_sum_of_tri(n - p, tri_list[:i + 1]) for i, p in enumerate(tri_list))
```

```python
for index in range(1, 15):
    t = tri(index)
    tri_list = [tri(i) for i in range(1, index+1)]
    print(count_sum_of_tri(t, tri_list))
```
