```python
from fractions import Fraction
from math import ceil, floor
from functools import cache
```

# $k = 1/a_1 + 1/a_2 + ... + 1/a_n$
We first implement a function that determines the number of ways to write $k \in \mathbb{Q}, k \geq 0$ as a sum of reciprocals of $n$ integers.

```python
@cache
def num_ways_sum_reciprocals(k, n, min_int: int = 1):
    if k == n == 0:
        return 1
    elif k <= 0 or n < 0:
        return 0
    max_int = floor(n / k)
    return sum(num_ways_sum_reciprocals(k - Fraction(1, d), n - 1, d) for d in range(min_int, max_int + 1))

num_ways_sum_reciprocals(Fraction(1, 1), 5)
```

```python
# Storing and returning the decompositions
def ways_sum_reciprocals(k, n, min_int: int = 1):
    if n == 0:
        return [[]] if k == 0 else []
    if k <= 0:
        return []
    max_int = floor(n / k)

    results = []
    for d in range(min_int, max_int + 1):
        remainder = k - Fraction(1, d)
        for rest in ways_sum_reciprocals(remainder, n - 1, min_int=d):
            results.append([Fraction(1, d)] + rest)
    return results

ways_sum_reciprocals(Fraction(1, 2), 3, 3)
```

```python
def greedy_egyptian_fraction_decomposition(k, min_denom=1, disallow_repeats=True):
    if k == 0:
        return []
    new_denom = max(ceil(1/k), min_denom)
    new_frac = Fraction(1, new_denom)
    new_k = k - new_frac
    new_min_denom = new_denom + disallow_repeats
    return [new_frac] + greedy_egyptian_fraction_decomposition(new_k, new_min_denom, disallow_repeats=disallow_repeats)
```

```python
greedy_egyptian_fraction_decomposition(Fraction(5, 121), min_denom=1, disallow_repeats=True)
```

### Finding different polygons whose combine perfectly at a corner

```python
# Given n polygons, examine the fraction n/2 - 1
all_ways = []
for n in range(3, 7):
    all_ways.extend(ways_sum_reciprocals(Fraction(n, 2) - 1, n, 3))

print(all_ways)
print(len(all_ways))
```
