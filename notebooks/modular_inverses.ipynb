# Modular Inverses

Given a positive integer $n,$ when do there exist integers $0 \leq a, b < n$ such that $a$ and $b$ are both additive and multiplicative inverses modulo $n?$

As it turns out, this holds if and only if $-1$ is a quadratic residue modulo $n.$ So, when can this happen?

Conjecture: it happens if and only if n can be written as a sum of coprime squares.

```python
from core.number_theory import is_nth_power
```

```python
def is_quadratic_residue(k, n):
    for i in range(0, n//2 + 1):
        if i**2 % n == (k % n):
            return True, i

    return False, 0
```

```python
[(n, x[1]) for n in range(1, 100) if (x := is_quadratic_residue(-1, n))[0]]
```
