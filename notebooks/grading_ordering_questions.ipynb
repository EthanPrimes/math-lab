```python
import itertools
import math
```

```python
def grade_kendall_tau(ranking):
    """Grading based on the minimum switches to get to ordered list.

    Computes the minimum number of consecutive switches needed to order
    the list and compares this to the worst-case ordering (reverse order).
    This is called the `Kendall tau distance`

    Args:
        ranking: List of unique integers.

    Returns:
        float representing the percentage of points the student gets.
    """
    curr_ranking = list(ranking).copy()
    n = len(ranking)
    switches = 0
    for _ in range(n - 1):
        for j in range(n - 1):
            if curr_ranking[j] > curr_ranking[j+1]:
                curr_ranking[j], curr_ranking[j+1] = (
                    curr_ranking[j+1],
                    curr_ranking[j]
                )
                switches += 1

    max_switches = n*(n - 1)/2

    return 1 - switches / max_switches
```

```python
N = 2
true_list = list(range(N))
perms = list(itertools.permutations(true_list))

sum(grade_kendall_tau(list(perm)) for perm in perms) / math.factorial(N)
```
