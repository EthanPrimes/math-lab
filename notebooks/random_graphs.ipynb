```python
from math import log

from matplotlib import pyplot as plt
import networkx as nx
import numpy as np
```

```python
G = nx.Graph()
G.add_node(1)
G.add_edge(1, 2)
print(G.nodes())
print(G.edges())
print(G.number_of_nodes(), G.number_of_edges())
```

```python
n = 10
p = log(n) / n / 1.2
G = nx.gnp_random_graph(n, p)

# Evaluating result
print(nx.is_connected(G))
print(nx.number_connected_components(G))
print(nx.connected_components(G))
nx.draw(G, node_size=20)
```

```python
n = 10**3
iters = 10**2
threshold = log(n) / n
p_vals = np.linspace(0, min(1, threshold*2), 10**2)
connected_prob = [sum(nx.is_connected(nx.gnp_random_graph(n, p)) / iters for _ in range(iters)) for p in p_vals]

plt.plot(p_vals, connected_prob)
plt.axvline(x=threshold, c="red", label=r"$\frac{ln(n)}{n}$ line")
plt.xlabel("p")
plt.ylabel("probability graph is fully connected")
plt.title(f"Probability That ER Graph With {n} Nodes is Fully Connected")
plt.legend()
plt.show()
```
