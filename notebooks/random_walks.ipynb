# 1-Dimensional Case
The probability of a random walk having touched the origin after $n$ unit steps asymptotically approaches $1 - \sqrt{\frac{2}{\pi n}}.$

Hence, we can use this to compute pi.

```python
import random
```

```python
def rand_step():
    return random.randint(0, 1) * 2 - 1

num_walks = 10000000
num_steps = 100
paths = [rand_step() for _ in range(num_walks)]

for i in range(num_steps - 1):
    paths = [p + rand_step() for p in paths]
    paths = [p for p in paths if p != 0]

print(len(paths) / num_walks)
print(1 / pow(3.14159 * num_steps / 2, 1/2))
```
