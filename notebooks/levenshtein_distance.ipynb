# Levenshtein Distance

The Levenshtein distance is the minimum number of insertsions, deletions, or character changes needed to transform one string into another. In this notebook we try to implement and then optimize a function that can compute this distance.

```python
from time import perf_counter as pf
import timeit
```

```python
ALPHABET = [chr(k) for k in range(97, 123)]
```

```python
def all_single_changes(str1: str) -> set[str]:
    """Returns a set of all single character modifications for str1.
    """
    n = len(str1)
    all_changes = set()
    for i in range(2*n + 1):
        ip = i//2
        if i % 2 == 0:
            all_changes.update([str1[:ip] + l + str1[ip:] for l in ALPHABET])
        else:
            all_changes.update([str1[:ip] + l + str1[ip + 1:] for l in ALPHABET])

    all_changes = {string.strip() for string in all_changes if string != str1}
    return all_changes


def levenshtein_distance(str1: str, str2: str) -> int:
    """Computes the Levenshtein distance between two strings.

    Naively determines the minimum number of changes needed to turn
    str1 into str 2.

    Args:
        str1: string input
        str2: string input

    Returns:
        An integer indicating the distance.

    Examples:
        >>> levenshtein_distance("sky", "fly")
        2
        >>> levenshtein_distance("abba", "abab")
        2
        >>> levenshtein_distance("", "test")
        4
    """
    n = 0
    related_strings = set([str1])
    while str2 not in related_strings:
        new_strings = set()
        for string in related_strings:
            new_strings.update(all_single_changes(string))

        n += 1
        related_strings = new_strings

    return n
```

```python
start = pf()
levenshtein_distance("", "aaaa")
print(pf() - start)
```
