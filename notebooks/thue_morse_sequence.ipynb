# Thue-Morse sequence
A010060

```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
def thue_morse(n):
    return int(bool(bin(n-1)[2:].count("1") % 2))

def compute_zero_sum(k):
    return sum([1/n for n in range(1, k+1) if not thue_morse(n)])

def compute_one_sum(k):
    return sum([1/n for n in range(1, k+1) if thue_morse(n)])
```

```python
N = 10000

x_vals = np.arange(1, N+1)
zero_sums = np.array([compute_zero_sum(i) for i in range(1, N+1)])
one_sums = np.array([compute_one_sum(i) for i in range(1, N+1)])

plt.plot(x_vals, zero_sums, label="Zero Sums")
plt.plot(x_vals, one_sums, label="One Sums")
plt.xlabel("N")
plt.ylabel("partial sums")
plt.legend()
plt.title("Comparing Reciprocal Thue-Morse Sums")
plt.show()

print(zero_sums - one_sums)
```
