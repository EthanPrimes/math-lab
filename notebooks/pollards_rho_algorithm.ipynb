## A notebook for testing / implementing Pollard's Rho Algorithm.

```python
import math
import random
from time import perf_counter as pf
from collections.abc import Callable
```

```python
def timing_decorator(func):
    def wrapper(*args, **kwargs):
        start = pf()
        res = func(*args, **kwargs)
        print(f"Function call time: {pf() - start}")
        return res

    return wrapper
```

```python
@timing_decorator
def find_factor(n: int) -> bool:
    i = 2
    up_bound = math.sqrt(n)
    while i <= up_bound:
        if n % i == 0:
            return i
        i += 1

    raise RuntimeError("The algorithm failed to find a factor.")

@timing_decorator
def rho_factor(n: int, x0: int = 2, c: int = 1) -> int:
    # Optimization for simple primes
    if math.gcd(n, 30) > 1:
        return math.gcd(n, 30)

    x, y, d = x0, x0, 1
    g = lambda v: (v**2 + c) % n
    while d == 1:
        x = g(x)
        y = g(g(y))
        d = math.gcd(abs(x - y), n)

    if d == n:
        # Failed to find a factor; randomize inputs and try again
        return rho_factor(n, random.randint(0, n - 1), random.randint(1, n - 2))
    else:
        return d
```

```python
# Timing trial division vs. rho algorithm
n = 2038074743 * 179424673
print(find_factor(n))
print(rho_factor(n))
```
