```python
import re
import sys
from time import perf_counter as pf
from core.primes import is_prime
```

```python
n = 104729
start = pf()
res = not re.match(r"^.?$|^(..+?)\1+$", "1" * n)
print("Regex Approach:")
print(f"Time: {pf() - start}")
print(f"Result: {res}\n")

start = pf()
res = is_prime(n)
print("Standard Approach:")
print(f"Time: {pf() - start}")
print(f"Result: {res}")
```
